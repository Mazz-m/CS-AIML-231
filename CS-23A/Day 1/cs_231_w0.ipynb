```python
from sklearn.tree import DecisionTreeClassifier
import numpy as np
```

Now, AI needs data to learn. Let's look at 10 days of weather right here in Mogadishu. We have two numbers for each day: the temperature in Celsius, and whether it rained (1 for yes, 0 for no).

```python
# The Data: 10 days of Mogadishu weather [Temp (°C), Is Raining?]
X = np.array([[30,0], [28,1], [32,0], [25,1], [35,0], [29,0], [24,1], [31,0], [26,1], [33,0]])
```

Next, we give the AI the 'answers'. For those 10 days, did you end up needing a jacket? (0 for no, 1 for yes). Because it's Mogadishu, you usually don't need one, unless it rained heavily or it was unusually cool!

```python
# The Labels: Did we need a jacket? (0 = No, 1 = Yes)
y = np.array([0, 1, 0, 1, 0, 0, 1, 0, 1, 0])
```

The "Magic" Trick

Now for the 'magic'. This single line is the AI learning. It looks at our data X, looks at the answers y, and uses basic statistics to find the hidden rules connecting them.

```python
# The "Magic" Trick: Train the model in ONE line of math!
model = DecisionTreeClassifier().fit(X, y)
```

The model has learned the rules. Now let's test it. Today is 27°C and it is raining. Based purely on the math it just calculated from our 10 days of data, what does it think we should do?

```python
# The Prediction: Today is 27°C and it's raining. Do we need a jacket?
print("Wear a jacket today?", "YES!" if model.predict([[30, 0]])[0] == 1 else "NO!")
```
